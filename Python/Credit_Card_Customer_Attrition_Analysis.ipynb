```python
df['Inactive_Risk'] = (df['Months_Inactive_12_mon'] >= 3).astype(int)
df['Transaction_Risk'] = (df['Total_Trans_Ct'] < 50).astype(int)
df['Contact_Risk'] = (df['Contacts_Count_12_mon'] >= 3).astype(int)

df['Risk_Score'] = ( df['Inactive_Risk'] + df['Transaction_Risk'] +  df['Contact_Risk'])

df[[
    'Months_Inactive_12_mon',
    'Total_Trans_Ct',
    'Contacts_Count_12_mon',
    'Risk_Score'
]].head(10)
```

```python
risky_customers = df[df['Risk_Score'] == 3]

risky_customers[[
    'Customer_Age',
    'Months_Inactive_12_mon',
    'Total_Trans_Ct',
    'Contacts_Count_12_mon',
    'Risk_Score'
]]

```

### Conclusions

1) The income group below $40k contains the largest number of customers and also the largest raw number of attrited    customers that is 256

2) Existing customer's average transaction amount is approximately 60% higher than attrited customers.
   Hence , attrited customers have substantially lower transaction amounts, indicating considerably lower credit-      card spending activity.

3) Existing customers make approximately 55% more transactions than attrited customers.Therefore attrited customers    are significantly less frequent users of the credit card.

4) Attrited customers have a lower average credit limit by about $905, indicating an association between lower         credit limits and attrition in this dataset.

5) 95% of the customer base uses Blue card
   Gold customers have the highest average transaction amount at approximately $2,851 , while PLatinum has the   lowest at $1,702. However, Platinum has only 3 customers, so its $1,702 average is not reliable enough to drive     a business decision

6) Developed a 0–3 customer risk score using inactivity, transaction count, and contact frequency, identifying customers with scores of 2–3 as high-risk for proactive retention.
